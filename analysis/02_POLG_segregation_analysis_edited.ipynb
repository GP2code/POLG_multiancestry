## Imports

```python
import pandas as pd
import statsmodels.api as sm
import seaborn as sns
import matplotlib.pyplot as plt
```

## Define paths

```python
# Read the CSV file into a pandas DataFrame

POLG_data = pd.read_csv('/content/sample_data/POLG_coding_vars.csv')
#Inspect the data
print(POLG_data.shape)
POLG_data.head()
```

### Group by family ID and extract zygosity

```python
#Check segregating variants (variant present in more than 1 individual in a single family) 
segregation_counts = pd.concat(g for _, g in POLG_data.groupby(['family_id', 'chr:pos:ref:alt']) if len(g) > 1)
segregation_counts_mode = pd.concat(g for _, g in POLG_data.groupby(['family_id', 'chr:pos:ref:alt','mode']) if len(g) > 1)
```

```python
segregation_counts.to_csv('segregation_counts.csv', index=True)
print(segregation_counts)
segregation_counts_mode.to_csv('segregation_counts_mode.csv', index=True)
print(segregation_counts_mode)
```
